```python
income_median = df["annual_income"].median()
employment_median = df["employment_years"].median()

df["annual_income"] = df["annual_income"].fillna(income_median)
df["employment_years"] = df["employment_years"].fillna(employment_median)
```

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound
```

```python
income_outliers, low, high = detect_outliers_iqr(df, "annual_income")
print(f"Range: {low} to {high}")
print(income_outliers[["customer_id", "annual_income"]])

loan_outliers, low2, high2 = detect_outliers_iqr(df, "loan_amount")
print(f"Range: {low2} to {high2}")
print(loan_outliers[["customer_id", "loan_amount"]])
```
